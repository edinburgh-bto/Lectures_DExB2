
**Question 1**
Which of the following best describes the null hypothesis in a significance test?

A) The hypothesis that there is no difference or effect in the population
B) The hypothesis that there is a large and important difference in the population
C) The hypothesis that your experimental treatment had a positive effect
D) The hypothesis that your sample data perfectly represent the population
E) The hypothesis that your statistical test will produce a significant result

**Correct answer: A**

---

**Question 2** 
What is the primary purpose of a Shapiro-Wilk test?

A) To determine if a dataset is compatible with being normally distributed 
B) To calculate the correlation between two numerical variables 
C) To compare the means of three or more groups 
D) To estimate the slope of a linear relationship 
E) To test if two categorical variables are associated

**Correct answer: A**

---

**Question 3**
Why is the arbitrary threshold of p < 0.05 criticised in modern statistics?

A) It guarantees that all findings are biologically significant
B) It encourages dichotomous thinking and can lead to over-interpretation of results near the threshold
C) It ensures that all published results are true positives
D) It eliminates the need for replication studies
E) It guarantees that p-values will always decrease with increased sample size

**Correct answer: B**

---
**Question 4**
In a study comparing plant growth under two light conditions, researchers report: t_13 = 3.2, p = 0.002. What does this indicate?

A) The data are compatible with the null hypothesis of no difference
B) The data are incompatible with the null hypothesis of no difference, suggesting a real difference in growth
C) The effect size is 3.2 mm
D) The result is not statistically significant
E) The sample size was too small to draw conclusions

**Correct answer: B**

**Question 5**
If a 95% confidence interval for the difference between two means includes zero, what can be concluded about the p-value?

A) The p-value must be greater than 0.05
B) The p-value must be less than 0.05
C) The p-value must be exactly 0.05
D) The p-value is unrelated to the confidence interval
E) The confidence interval and p-value will always contradict each other

**Correct answer: A**

**Question 6**

A) The observed difference in means 
B) The value of the t-statistic and degrees of freedom 
C) The p-value 
D) The sample size

**Question 8** 
In an ANOVA test comparing means across three groups, a significant result (p < 0.05) indicates what?

A) All three groups have significantly different means 
B) At least one group has a mean that is significantly different from the others 
C) The means of all groups are exactly equal
